```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('train_and_test2.csv')
```

```python
df.drop(columns=df.filter(regex=r"^zero").columns, inplace=True)
```

```python
df.head()
```

# 1. Categorical Data

```python
df['Pclass'].value_counts().plot(kind='pie', autopct ='%.2f')
```

# 2. Numerical Data

## a. Histogram

```python
import matplotlib.pyplot as plt

plt.hist(df['Age'],bins=50)
```

## b. **Distplot**
### Shows probability through moving graph instad of exact values and if there is skewness in data.

```python
sns.distplot(df['Age'])
```

## c. Boxplot
### Gives 5 number summary and helps to detect noise and outliers


```python
sns.boxplot(df['Age'])
```
